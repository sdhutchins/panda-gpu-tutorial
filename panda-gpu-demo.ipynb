# Panda GPU Demo

This is a demo of using netzoopy's gpu capabilities to run Panda.

```python
# Core imports
import numpy as np
import pandas as pd
import s3fs
import psutil
import os

# GPU
import cupy as cp

# Test that everything is properly set up
print("NumPy:", np.__version__)
print("CuPy:", cp.__version__)
print("GPU count:", cp.cuda.runtime.getDeviceCount())

# netZoo
import netZooPy
from netZooPy.panda import Panda
from netZooPy.lioness import Lioness
```

## Download data

First, we get the data from the GRAND database. This can be done directly through pandas.

```python
dirs = ["data", "output"]
for d in dirs:
    os.makedirs(d, exist_ok=True)

LCL_ppi = pd.read_csv("https://granddb.s3.amazonaws.com/cells/ppi/LCL_ppi.txt", sep="\t", header=None)
LCL_expression = pd.read_csv("https://granddb.s3.amazonaws.com/optPANDA/expression/Hugo_exp1_lcl.txt", sep="\t", header=None)
LCL_motif = pd.read_csv("https://granddb.s3.amazonaws.com/gpuPANDA/Hugo_motifCellLine_reduced.txt", sep="\t", header=None)

LCL_ppi.to_csv("data/LCL_ppi.txt", sep="\t", index=False, header=False)
LCL_expression.to_csv("data/Hugo_exp1_lcl.txt", sep="\t", index=False, header=False)
LCL_motif.to_csv("data/Hugo_motifCellLine_reduced.txt", sep="\t", index=False, header=False)
```

```python
# Set data variables
expression_data = "data/Hugo_exp1_lcl.txt"
motif_data = "data/Hugo_motifCellLine_reduced.txt"
ppi_data = "data/LCL_ppi.txt"
```

## Run Panda with GPU & precision flags
The precision flags allow to compute the network in single (7 decimal digits) or double precision (15 decimal digits). Although double precision networks are more accurate, single precision can be useful to free memory on devices and to go faster. It is a tradeoff that depends on the application

```python
panda_obj = Panda(expression_data, motif_data, ppi_data, computing='gpu',
                  precision='double', save_tmp=False, save_memory = False,
                  remove_missing=True, keep_expression_matrix = True, modeProcess = 'intersection')
```

```python
panda_obj.save_panda_results('output/double_gpu_panda.txt')
```
